# Supervised Learning – Regression & Linear Regression Demo

Predict numeric values using Linear Regression.
What Problem Are We Solving?

👉 You want to predict:

House price

Salary

Marks

All these are NUMBERS.

This is REGRESSION.

🔹 What is Linear Regression?

Linear Regression means:

“Draw a straight line that best fits the data.”

That line is used for prediction.

🔹 Why Do We Need This Line?

Because:

New data will come

We cannot guess price manually

Line gives best possible estimate

🔹 How Linear Regression Works (Very Simple)

1️⃣ Look at all points
2️⃣ Draw a straight line
3️⃣ Line minimizes error
4️⃣ Use line to predict new value

image.png

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
import numpy as np
```

```python
x = np.array([800,1000,1200,1500]).reshape(-1,1) # house size
y = np.array([40,50,65,80]) # price(output)
x
```

```python
model = LinearRegression()
model.fit(x,y) #model learns relationship
```

```python
plt.scatter(x,y,color = 'blue')
plt.plot(x,model.predict(x),color = 'red')
plt.xlabel('House Size')
plt.ylabel('Price')
plt.title('House Price Prediction')
plt.show()
```
