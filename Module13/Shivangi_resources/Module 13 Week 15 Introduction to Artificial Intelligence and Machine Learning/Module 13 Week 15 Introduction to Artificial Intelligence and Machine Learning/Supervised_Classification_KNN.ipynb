
# Supervised Learning – Classification (KNN Algorithm)

## Objective
Classify a person as **Fit** or **Unfit** based on **Height** and **Weight**.

This is a **Supervised Learning** problem because:
- Output labels are known
- Output is a category


```python

import numpy as np
import matplotlib.pyplot as plt
from sklearn.neighbors import KNeighborsClassifier
```


## Step 1: Create Dataset
- X → input features
- y → output labels


```python

X = np.array([
    [160, 55],
    [170, 65],
    [165, 85],
    [175, 90],
    [180, 70],
    [155, 48]
])

y = np.array([
    "Fit",
    "Fit",
    "Unfit",
    "Unfit",
    "Fit",
    "Fit"
])
```


## Step 2: Check Shapes
X must be 2D and y must be 1D.


```python

X.shape, y.shape
```


## Step 3: Visualize Training Data


```python

for label in np.unique(y):
    plt.scatter(
        X[y == label][:, 0],
        X[y == label][:, 1],
        label=label
    )

plt.xlabel("Height (cm)")
plt.ylabel("Weight (kg)")
plt.title("Training Data (Fit vs Unfit)")
plt.legend()
plt.show()
```


## Step 4: Create KNN Classifier
K = 3 nearest neighbors.


```python

knn = KNeighborsClassifier(n_neighbors=3)
```


## Step 5: Train the Model


```python

knn.fit(X, y)
```


## Step 6: Predict for New Person


```python

new_person = np.array([[168, 70]])
knn.predict(new_person)
```


## Step 7: Visualize Prediction


```python

for label in np.unique(y):
    plt.scatter(
        X[y == label][:, 0],
        X[y == label][:, 1],
        label=label
    )

plt.scatter(
    new_person[0][0],
    new_person[0][1],
    color='red',
    s=200,
    marker='X',
    label='New Person'
)

plt.xlabel("Height (cm)")
plt.ylabel("Weight (kg)")
plt.title("KNN Classification Result")
plt.legend()
plt.show()
```


## Key Takeaways
- This is supervised learning
- Output is categorical
- KNN uses distance and majority voting
