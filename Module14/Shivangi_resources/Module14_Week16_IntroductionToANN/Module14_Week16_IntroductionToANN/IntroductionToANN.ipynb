```python
# ===============================
# Regression using ANN
# ===============================

import numpy as np
import pandas as pd
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

# Create dummy dataset
np.random.seed(23)

X = np.random.rand(1000, 3)
'''
x1  x2   x3
0.3  0.3  0.4
0.4  0.3  0.2
'''

y = (X[:,0]*50000 + X[:,1]*30000 + X[:,2]*20000) + 10000

# Split data
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

# Scale
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

# Build model
model = keras.Sequential([
    layers.Dense(16, activation='relu', input_shape=(3,)), # 16 neuron, 5 weights, 1 bias (3x16)+16 =  64 parameters
    layers.Dense(8, activation='relu'), #8 neuron, 16 outputs from previous layer (16x8)+8 = 136
    layers.Dense(1) # 0.82=82% -> probability class : 1
])

# Compile
model.compile(optimizer='adam',
              loss='mse',
              metrics=['mae'])

# Train
history = model.fit(X_train, y_train,
                    epochs=50,
                    validation_split=0.2)
'''
During epochs:
1) forward propagation
2) compute predictions
3) compute loss
4) backpropagation
5) update weights
6) repeat
'''

# Evaluate
loss, mae = model.evaluate(X_test, y_test)
print("Test MAE:", mae)

# Predict
predictions = model.predict(X_test[:5])
print(predictions)
```

```python
# ===============================
# Classification using ANN
# ===============================

from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers

# Create synthetic dataset
X, y = make_classification(n_samples=1000,
                           n_features=10,
                           n_classes=2,
                           random_state=42)

# Split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

# Scale
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

# Build model
model = keras.Sequential([
    layers.Dense(32, activation='relu', input_shape=(10,)),
    layers.Dense(16, activation='relu'),
    layers.Dense(1, activation='sigmoid')
])

# Compile
model.compile(optimizer='adam',
              loss='binary_crossentropy',
              metrics=['accuracy'])

# Train
history = model.fit(X_train, y_train,
                    epochs=50,
                    validation_split=0.2)

# Evaluate
loss, accuracy = model.evaluate(X_test, y_test)
print("Test Accuracy:", accuracy)

# Predict
predictions = model.predict(X_test[:5])
print(predictions)
```
